```python
import csv
from collections import defaultdict
```

```python
data = []
with open('datasets/Brand_Sales_AdSpend_Data.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        data.append(row)
```

```python
countries = set([r['Country'] for r in data])
countries
```

```python
spend = defaultdict(float)
for r in data:
    spend[r['Country']] += float(r['Total Ad Spend\xa0'])
```
